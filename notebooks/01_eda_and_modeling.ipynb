# Exploratory Data Analysis and Modeling

**Educational screening project only — not medical diagnosis.** Run `python -m src.train_model` first to generate the cleaned dataset. This notebook is intentionally readable and leaves figures reproducible rather than embedding invented findings.

````python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

ROOT = Path.cwd().resolve().parent if Path.cwd().name == 'notebooks' else Path.cwd()
data = pd.read_csv(ROOT / 'data' / 'processed' / 'clean_heart_disease.csv')
data.shape, data.isna().sum().sum(), data.duplicated().sum()
````

## Target and demographic distributions

Observe class balance and whether apparent sex differences deserve further analysis. They are descriptive patterns only, not causal evidence.

````python
fig, axes = plt.subplots(1, 3, figsize=(16, 4))
sns.countplot(data=data, x='target', ax=axes[0]); axes[0].set_title('Target distribution')
sns.histplot(data=data, x='age', hue='target', kde=True, ax=axes[1]); axes[1].set_title('Age by target')
sns.countplot(data=data, x='sex', hue='target', ax=axes[2]); axes[2].set_title('Sex vs target')
plt.tight_layout()
````

**Observation:** Execute the chart and summarize the observed class balance and distribution differences here. Avoid treating these patterns as clinical conclusions.

````python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.countplot(data=data, x='cp', hue='target', ax=axes[0]); axes[0].set_title('Chest pain type vs target')
sns.heatmap(data.corr(numeric_only=True), cmap='vlag', center=0, ax=axes[1]); axes[1].set_title('Correlation heatmap')
plt.tight_layout()
````

## Continuous and categorical feature checks

Inspect distribution overlap, outliers, and category counts before interpreting model behavior.

````python
continuous = ['chol', 'trestbps', 'thalach', 'oldpeak']
fig, axes = plt.subplots(2, 4, figsize=(16, 8))
for ax, col in zip(axes[0], continuous):
    sns.histplot(data=data, x=col, hue='target', kde=True, ax=ax)
for ax, col in zip(axes[1], continuous):
    sns.boxplot(data=data, x='target', y=col, ax=ax)
plt.tight_layout()

categorical = ['fbs', 'restecg', 'exang', 'slope', 'ca', 'thal']
fig, axes = plt.subplots(2, 3, figsize=(14, 8))
for ax, col in zip(axes.flat, categorical):
    sns.countplot(data=data, x=col, hue='target', ax=ax)
plt.tight_layout()
````

**Observation:** Record notable variation, sparseness, and possible data-quality concerns after running the plots. Small subgroup counts should not be over-interpreted.
